```python
from google.colab import files
uploaded = files.upload()
zip_name = next(iter(uploaded))
!unzip -q -o "$zip_name"
import glob, os
# the project folder: the one with src/autopilot.py (works for the project zip or the whole repository)
os.chdir(os.path.dirname(os.path.dirname(os.path.abspath(sorted(glob.glob('**/src/autopilot.py', recursive=True), key=len)[0]))))
print('Project folder:', os.getcwd())
```
