# Self-driving car: training on Google Colab

**No GPU needed:** the network is small, and most of the time goes to the driving simulation, which runs on the CPU. You can keep the regular runtime (CPU).

At the end, download `models.zip` and extract it into the project folder on your computer, then run the app there.

## 1. Upload the project
Choose the project zip file.

```python
from google.colab import files
uploaded = files.upload()
zip_name = next(iter(uploaded))
!unzip -q -o "$zip_name"
import glob, os
# the project folder: the one with src/validation.py (works for the project zip or the whole repository)
os.chdir(os.path.dirname(os.path.dirname(os.path.abspath(sorted(glob.glob('**/src/validation.py', recursive=True), key=len)[0]))))
print('Project folder:', os.getcwd())
```

## 2. Training
16 cars drive in parallel on random tracks. Every 2,500 steps the network is tested on 10 validation tracks, and the best one is saved to `models/driver.pt`. A good driver reaches about 3 laps in 30 seconds.

```python
!python -m src.train
```
